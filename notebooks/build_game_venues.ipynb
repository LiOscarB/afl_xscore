```python
import pandas as pd
```

```python
fixtures = pd.read_csv('../data/fixtures.csv')
venues = pd.read_csv('../data/venues.csv')

game_venues = fixtures.merge(venues, on='venue', how='left', validate='many_to_one')
assert len(game_venues) == len(fixtures)
game_venues.head()
```

```python
# Venues in fixtures with no dimensions
game_venues[game_venues['venue_x'].isna()]['venue'].value_counts()
```

```python
game_venues.to_csv('../data/game_venues.csv', index=False)
```
