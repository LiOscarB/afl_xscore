```python
import pandas as pd
import requests
```

```python
# Fixtures from the Squiggle API (https://api.squiggle.com.au)
HEADERS = {'User-Agent': 'afl_xscore research (shaqeelsahrizan@gmail.com)'}
SEASONS = range(2021, 2027)

games = []
for year in SEASONS:
    resp = requests.get('https://api.squiggle.com.au/', params={'q': 'games', 'year': year}, headers=HEADERS, timeout=30)
    resp.raise_for_status()
    games += resp.json()['games']
len(games)
```

```python
# Use the same team names as the shots data
TEAM_MAP = {'Brisbane Lions': 'Brisbane'}

fixtures = pd.DataFrame({
    'season': [g['year'] for g in games],
    'round': [g['round'] for g in games],
    'home_team': [TEAM_MAP.get(g['hteam'], g['hteam']) for g in games],
    'away_team': [TEAM_MAP.get(g['ateam'], g['ateam']) for g in games],
    'venue': [g['venue'] for g in games],
    'date': [g['date'][:10] for g in games],
}).sort_values(['season', 'round', 'date', 'home_team'])
fixtures.head()
```

```python
fixtures.to_csv('../data/fixtures.csv', index=False)
```
