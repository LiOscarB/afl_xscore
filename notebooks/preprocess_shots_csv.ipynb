```python
import pandas as pd 
import numpy as np
```

```python
shots_df = pd.read_csv('../data/afl_shots.csv')
shots_df.head()
```

```python
jez_df = shots_df[shots_df['player_id']== '230055']
jez_df[(jez_df['Season'] == 2022) & (jez_df['Round'] == 18) & (jez_df['xScore'] == 1.3)].head()
```

```python
jez_df['dist_to_goal'] = np.sqrt((jez_df['x'])**2 + (jez_df['y'])**2)
jez_df['dist_to_l_post'] = np.sqrt((jez_df['x'])**2 + (3.2 - jez_df['y'])**2)
jez_df['dist_to_r_post'] = np.sqrt((jez_df['x'])**2 + (-3.2 - jez_df['y'])**2)
jez_df['angle_to_goal'] = (np.arccos((jez_df['dist_to_l_post']**2 + jez_df['dist_to_r_post']**2 - 6.4**2) / (2 * jez_df['dist_to_l_post'] * jez_df['dist_to_r_post'])) * (180/np.pi)
)
jez_df[(jez_df['Season'] == 2022) & (jez_df['Round'] == 18) & (jez_df['xScore'] == 1.3)].head()
```

```python
shots_df['dist_to_goal'] = np.sqrt((shots_df['x'])**2 + (shots_df['y'])**2)
shots_df['dist_to_l_post'] = np.sqrt((shots_df['x'])**2 + (3.2 - shots_df['y'])**2)
shots_df['dist_to_r_post'] = np.sqrt((shots_df['x'])**2 + (-3.2 - shots_df['y'])**2)
shots_df['angle_to_goal'] = (np.arccos((shots_df['dist_to_l_post']**2 + shots_df['dist_to_r_post']**2 - 6.4**2) / (2 * shots_df['dist_to_l_post'] * shots_df['dist_to_r_post'])) * (180/np.pi)
)
shots_df['set_shot_flag'] = np.where(shots_df["ShotType"] == 'Set Position', 1, 0)
shots_df_clean = shots_df.drop(columns = ['Distance', 'Pocket','dist_to_l_post', 'dist_to_r_post', 'ShotType', 'Unnamed: 0'])
shots_df_clean.head()
```

```python
shots_df_clean.to_csv('../data/processed_shots.csv', index=False)
```
